## The difference between System.out.println and System.out.print

```java
System.out.println("Hello");
System.out.println("Hello");
System.out.println("Hello");
```

```java
System.out.print("Hello");
System.out.print("Hello");
System.out.print("Hello");
```

After printing the text (String) in the brackets, System.out.println moves the cursor onto the next line, whereas System.out.print does not. This means in the latter a sequence of things printed go on the same line. In the former they each start on different lines.
