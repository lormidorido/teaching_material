### whatsYourName

#### What it does
This method asks the users name then greets them by name in a jokey way, all via the console window.

#### Implementation (how it works)
It uses a scanner to do input, reads in a single line of input and stores it in variable name to be printed later.


```java
// Asks the users name then greets them by name
//
public static void whatsYourName ()
{
     String name;
     Scanner scanner = new Scanner(System.in);
    
     System.out.println("What is your name?");
     name = scanner.nextLine();
     System.out.println("Hi " + name + ".");
     System.out.println("It's great for you to meet me.");
     return;
}
```

#### Testing

```java
whatsYourName();
```

### WhatsYourAge

#### What it does
This method asks the age of the person and returns the result.

#### Implementation (how it works)
It creates a scanner, prints the request for the age, reads in the next line typed as a string, converts it to an integer and returns that integer.


```java
// Asks the users age then greets them by name
//
public static int whatsYourAge ()
{
    String ageString;
    int age;
    Scanner scanner = new Scanner(System.in);
    
    System.out.println("How old are you?");
    ageString = scanner.nextLine();
    age = Integer.parseInt(ageString);
    return age;
}
```

#### Testing

```java
whatsYourAge ();
```

### ChatbotLevel2

#### What it does
This method prints a welcome message then asks for the user's name and greets them by name. It also asks for their ages and compare their age to its.

#### Implementation (how it works)
It first calls chattyHello to do a basic welcome, then calls whatsYourName to get the person's name and do the greeting. Then it calls whatsYourAge to get their age comparing it with its own age and using Math.abs to change a negative answer if any to a positive answer so that the number of years different is always positive. Finally it prints a message about that difference as stored in variable age_difference.


```java
// A chatbot - prints a welcome message, asks the user's name then greets them personally
// It then asks for their age and compares their age to its age
//
public static void ChatbotLevel2 ()
{
    final int myAge = 21;
    int age;
    int age_difference;
    
    chattyHello();
    whatsYourName();
    
    age = whatsYourAge ();
    age_difference = Math.abs(myAge - age);
    
    System.out.println("That means we are " + age_difference + " years apart.");
        
    return;
}
```
